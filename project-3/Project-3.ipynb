### Load the dataset

```python
data = pd.read_csv('cleaned_dataset.csv')
```

```python
# load the dataset into sql
connect = sqlite3.connect(":memory:")

data.to_sql(
    "data",
    connect,
    index=False
)
```

## Basic Select Queries

```python
# display all records
query1 = """
SELECT *
FROM data
"""
records = pd.read_sql_query(query1, connect)
records
```

```python
# selecting columns with important info
query2 = """
SELECT
    OrderID,
    Date,
    Product,
    Quantity,
    UnitPrice,
    TotalPrice
FROM data
"""
pd.read_sql_query(query2, connect)
```

```python
# display customer and order information
query3 = """
SELECT 
    CustomerID,
    Product,
    Quantity,
    OrderStatus,
    TotalPrice
FROM data
"""
pd.read_sql_query(query3, connect)    
```

```python
# display pricing information
query4 = """
SELECT
    Product,
    Quantity,
    UnitPrice,
    TotalPrice,
    CalculatedTotal,
    PriceDifference
FROM data;
"""

pd.read_sql_query(query4, connect)
```

## WHERE and ORDER BY to filter data & sort data

```python
# orders with quantity more than three
query5 = """
SELECT *
FROM data
WHERE Quantity > 3;
"""
pd.read_sql_query(query5, connect)
```

```python
# orders with more than 4 items in cart
query6 = """
SELECT
    OrderID,
    Product,
    Quantity,
    OrderStatus,
    ItemsInCart,
    TotalPrice
FROM data
WHERE ItemsInCart > 4
ORDER BY TotalPrice DESC;
"""
pd.read_sql_query(query6, connect)
```

```python
# display delivered orders and ones with totalprice > 1000, ordered by total price
query7 = """
SELECT
    OrderID,
    Product,
    Quantity,
    UnitPrice,
    TotalPrice
FROM data
WHERE 
    TotalPrice > 1000 AND 
    OrderStatus = "Delivered"
ORDER BY TotalPrice DESC;
"""
pd.read_sql_query(query7, connect)
```

```python
# pricing difference table
difference_query = """
SELECT
    OrderId,
    Product,
    TotalPrice,
    CalculatedTotal,
    PriceDifference
FROM data
WHERE PriceDifference <> 0
ORDER BY ABS(PriceDifference) DESC;
"""
pricing_difference = pd.read_sql_query(difference_query, connect)
pricing_difference
```

```python
# display orders that used debitcard as payment method ordered by the items in cart
query7 = """
SELECT
    OrderID,
    Product,
    Quantity,
    UnitPrice,
    TotalPrice
FROM data
WHERE PaymentMethod = "Debit Card"
ORDER BY TotalPrice DESC;
"""
pd.read_sql_query(query7, connect)
```

```python
# display highest value orders that were cancelled
query8 = """
SELECT 
    OrderID,
    Product,
    Quantity,
    TotalPrice
FROM data
WHERE OrderStatus = "Cancelled"
ORDER BY TotalPrice DESC;
"""
pd.read_sql_query(query8, connect)
```

## Basic Aggregation Functions

```python
# dataset overview
query9 = """
SELECT
    COUNT(*) AS Total_Orders
FROM data;
"""
overview = pd.read_sql_query(query9, connect)
overview
```

```python
# remove duplicated in customer id and return unique number of customers
query10 = """
SELECT 
    COUNT(DISTINCT CustomerID) AS Unique_Customers
FROM data;
"""
pd.read_sql_query(query10, connect)
```

```python
# display overall performance
performance_query = """
SELECT
    SUM(TotalPrice) AS Total_Revenue,
    AVG(TotalPrice) AS Average_Order_Value,
    SUM(Quantity) AS Total_Units_Sold
FROM data;
"""
overall_performance = pd.read_sql_query(performance_query, connect)
overall_performance
```

```python
# display product performance
query11 = """
SELECT
    Product,
    COUNT(*) AS Number_Of_Orders,
    SUM(Quantity) AS Units_Sold,
    SUM(TotalPrice) AS Revenue,
    AVG(TotalPrice) AS Average_Order_Value
FROM data
GROUP BY product
ORDER BY Revenue DESC;
"""
product_performance = pd.read_sql_query(query11, connect)
product_performance
```

```python
# display customer performance
query12 = """
SELECT
    CustomerId,
    COUNT(*) AS Number_Of_Orders,
    SUM(TotalPrice) AS Amount_Spent
FROM data
GROUP BY CustomerId
ORDER BY Amount_Spent DESC;
"""
customer_performance = pd.read_sql_query(query12, connect)
customer_performance
```

```python
# Referral performance
referral_query = """
SELECT
    ReferralSource,
    COUNT(*) AS Number_Of_Orders,
    SUM(TotalPrice) AS Revenue,
    AVG(TotalPrice) AS Average_Order_Value
FROM data
GROUP BY ReferralSource
ORDER BY Revenue DESC;
"""
referral_performance = pd.read_sql_query(referral_query, connect)
referral_performance
```

```python
# performance for all payment methods
payment_query = """
SELECT
    PaymentMethod,
    COUNT(*) AS Number_Of_Orders,
    SUM(TotalPrice) AS Revenue,
    AVG(TotalPrice) AS Average_Order_Value
FROM data
GROUP BY PaymentMethod
ORDER BY Revenue DESC;
"""
payment_method_performance = pd.read_sql_query(payment_query, connect)
payment_method_performance
```

```python
# order status performance summary
query13 = """
SELECT
    OrderStatus,
    COUNT(*) AS Number_Of_Orders,
    SUM(TotalPrice) AS Revenue
FROM data
GROUP BY OrderStatus
ORDER BY Revenue DESC;
"""
order_status_performance = pd.read_sql_query(query13, connect)
order_status_performance
```

```python
# Revenue generated by completed orders
query14 = """
SELECT
    Product,
    COUNT(*) AS Number_Of_Orders,
    SUM(TotalPrice) AS Revenue,
    AVG(TotalPrice) AS Average_Order_Revenue
FROM data
WHERE OrderStatus = "Delivered"
GROUP BY Product
ORDER BY Revenue DESC
"""
completed_orders_summary = pd.read_sql_query(query14, connect)
completed_orders_summary
```
